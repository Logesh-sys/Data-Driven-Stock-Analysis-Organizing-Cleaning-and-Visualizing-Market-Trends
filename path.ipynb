```python
base_folder = r"D:\Logesh\Data-Driven Stock Analysis Organizing, Cleaning, and Visualizing Market Trends\Data_Drivenstock"

all_data = []

for month in os.listdir(base_folder):
    month_path = os.path.join(base_folder, month)

    if os.path.isdir(month_path):

        for file in os.listdir(month_path):
            if file.endswith(".yaml") or file.endswith(".yml"):

                file_path = os.path.join(month_path, file)

                with open(file_path, "r") as f:
                    data = yaml.safe_load(f)

                print("Reading:", file_path)

                if isinstance(data, list):
                    all_data.extend(data)
                elif isinstance(data, dict):
                    all_data.append(data)

df = pd.DataFrame(all_data)

print("Loaded Successfully")
print("Rows:", len(df))
print(df.head())
```

2. Cumulative Return Over Time

```python
vol_df .head()
```

```python
# Convert date column to datetime
vol_df ['date'] = pd.to_datetime(vol_df ['date'])

# Sort values
cum_df = vol_df .sort_values(['Ticker', 'date'])

# Calculate cumulative return for each stock
cum_df['cumulative_return'] = (
    (1 + cum_df['daily_return'])
    .groupby(cum_df['Ticker'])
    .cumprod() - 1
)
```

```python
cum_df.head()
```

```python
#Top 5 performing stocks
top5 = (
    cum_df.groupby('Ticker')['cumulative_return']
    .last()
    .sort_values(ascending=False)
    .head(5)
)
top5_tickers = top5.index.tolist()
top5_tickers
```

```python
top5_data = cum_df[cum_df['Ticker'].isin(top5_tickers)]
top5_data
```

```python
plt.figure(figsize=(12,6))

for ticker in top5_tickers:
    stock = top5_data[top5_data['Ticker'] == ticker]

    plt.plot(
        stock['date'],
        stock['cumulative_return'],
        label=ticker
    )

plt.title("Cumulative Return for Top 5 Performing Stocks")
plt.xlabel("Date")
plt.ylabel("Cumulative Return")
plt.legend()
plt.grid(True)
```

5. Top 5 Gainers and Losers (Month-wise):


```python
stock_df['date'] = pd.to_datetime(stock_df['date'])
stock_df['month'] = stock_df['date'].dt.to_period('M')
```

```python
monthly_returns = (
    stock_df.sort_values('date')
    .groupby(['Ticker', 'month'])
    .agg(
        first_price=('close', 'first'),
        last_price=('close', 'last')
    )
    .reset_index()
)

monthly_returns['monthly_return'] = (
    (monthly_returns['last_price'] - monthly_returns['first_price'])
    / monthly_returns['first_price']
)
```

```python
top_gainers = (
    monthly_returns.sort_values(['month', 'monthly_return'], ascending=[True, False])
    .groupby('month')
    .head(5)
    .reset_index(drop=True)
)

top_losers = (
    monthly_returns.sort_values(['month', 'monthly_return'], ascending=[True, True])
    .groupby('month')
    .head(5)
    .reset_index(drop=True)
)
```

```python
months = monthly_returns['month'].unique()

for m in months:
    plt.figure(figsize=(10,5))

    gain = top_gainers[top_gainers['month'] == m]
    loss = top_losers[top_losers['month'] == m]

    plt.bar(gain['Ticker'], gain['monthly_return'], label='Top Gainers')
    plt.bar(loss['Ticker'], loss['monthly_return'], label='Top Losers')

    plt.title(f"Top 5 Gainers & Losers - {m}")
    plt.xlabel("Ticker")
    plt.ylabel("Monthly Return")
    plt.xticks(rotation=45)
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python
monthly_returns.to_csv("monthly_returns.csv", index=False)
top_gainers.to_csv("top_gainers.csv", index=False)
top_losers.to_csv("top_losers.csv", index=False)
```
